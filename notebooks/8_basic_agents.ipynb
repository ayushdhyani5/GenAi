```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_number(a:int,b:int) :
    """
        it will return the sum of two numbers
        Args: 
            a->number1
            b->number2
    """
    return a+b
@tool
def sub_number(a:int,b:int) :
    """
        it will return the diff of two numbers
        Args: 
            a->number1
            b->number2
    """
    return a-b
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
agent=create_agent(model=llm,tools=[add_number,sub_number],
    system_prompt="you are a math professor and always use tools for calculations"
)
```

```python

res=agent.invoke({
    "messages":[{"role":"user","content":"whats 766 + 673 - 888"}]
})
print(res["messages"][-1].content[0]["text"])
```
