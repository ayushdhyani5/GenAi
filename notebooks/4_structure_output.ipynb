```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm=ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
from pydantic import BaseModel,Field
from typing import List
class ResponseStructure(BaseModel):
    name:str=Field(description="Name of the user")
    email:str=Field(description="Email of the user")
    age:int=Field(description="Age of the user")

structure_llm=llm.with_structured_output(ResponseStructure)
```

```python
text = "My name is Ayush. My email is ayushdhyani764@gmail.com. I am 21 years old."

res = structure_llm.invoke(
    f"Extract the name, email, and age from this text: {text}"
)

print(res)
print(res.name)
print(res.email)
print(res.age)
res.model_dump()
```

```python
class movie(BaseModel):
    title:str=Field(description="movie title")
    year:int=Field(description="release year")
class AllMovies(BaseModel):
    movies:List[movie]=Field(description="")


movie_llm=llm.with_structured_output(AllMovies)
res=movie_llm.invoke(
     "give me 8 movies"
)
```

```python
c=len(res.movies)
for i in range (c):
    print(res.movies[i])
```
